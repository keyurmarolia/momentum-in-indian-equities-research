## Tax interpretation

FIFO gains, deductible trade expenses excluding STT, dated listed-equity rates, annual gain/loss netting, eight-year loss carryforwards, exemptions, grandfathering and cess produce a research estimate. Each annual payment reduces the modeled capital; later realized gains are scaled to the capital remaining while fixed rupee exemptions remain fixed. Cash-sleeve realization tax, surcharge and investor-specific adjustments are excluded. Results are a portfolio-level tax overlay, not an investor tax return.

Raw → after costs → post-tax comparisons in the strategy notebooks use these same assumptions.
